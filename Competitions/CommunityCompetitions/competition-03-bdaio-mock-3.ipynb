## 04 Data Preprocessing

```python
train.isna().sum()
```

```python
train.color_score.median(), train.color_score.mean()
```

```python
train.color_score = train.color_score.fillna(5.9)
```

```python
train.firmness_score = train.firmness_score.fillna(5.0)
```

```python
test.firmness_score.median(), test.firmness_score.mean()
```

```python
test.firmness_score = test.firmness_score.fillna(5.0)
```

```python
# one encoding :
one_hot_train = pd.get_dummies(train, columns=['variety'])
one_hot_test = pd.get_dummies(test, columns=['variety'])
```

```python
train.isna().sum()
```

```python
test.isna().sum()
```

```python
# features and target dividing :
X = one_hot_train.drop(columns='brix')
y = one_hot_train.brix
```

```python
# Standardization ;
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_train = scaler.fit_transform(X)
scaled_test  = scaler.fit_transform(one_hot_test)
```

```python
# dataset splittin :
from sklearn.model_selection import train_test_split

X_scaled , X_val , y_scaled, y_val = train_test_split(scaled_train, y, test_size=0.2, random_state= 19)
X_scaled_n, X_val_n ,  y_scaled_n, y_val_n = train_test_split(X, y, test_size=0.2, random_state= 19)
```

---
## 05 Model Building

```python
# linear Regression
from sklearn.linear_model import LinearRegression, Ridge

ln = LinearRegression()
rg = Ridge()

ln.fit(X_scaled, y_scaled)
ln_scl_pred = ln.predict(X_val)
```

```python
rg.fit(X_scaled, y_scaled)
rg_scl_pred = rg.predict(X_val)
```

```python
ln.fit(X_scaled_n, y_scaled_n)
ln_scln_pred = ln.predict(X_val)
```

```python
rg.fit(X_scaled_n, y_scaled_n)
rg_scln_pred = rg.predict(X_val)
```

## 06 Evaluation 

```python
from sklearn.metrics import root_mean_squared_error as rmse

print('Linear Regression Accuricy without scale :',  rmse(y_val, ln_scln_pred))
print('Linear Regression Accuricy with    scale :',  rmse(y_val, ln_scl_pred))
print()
print('Ridge Regression Accuricy without scale :',  rmse(y_val, rg_scln_pred))
print('Ridge Regression Accuricy with    scale :',  rmse(y_val, rg_scl_pred))
```

---
## 07 Final Prediction

```python
ln.fit(scaled_train, y)
final_pred_ln = ln.predict(scaled_test)

rg.fit(scaled_train, y)
final_pred_rg = rg.predict(scaled_test)
```

---
## 08 Submission

```python
# Linear Regression
ln_submission = pd.DataFrame({
    'id'   : test.id,
    'brix' : final_pred_ln
})

# Ridge Regression
rg_submission = pd.DataFrame({
    'id'   : test.id,
    'brix' : final_pred_rg
})
```

```python
ln_submission.to_csv('linear_submission.csv', index=False)
rg_submission.to_csv('Ridge_submission.csv', index=False)
```
