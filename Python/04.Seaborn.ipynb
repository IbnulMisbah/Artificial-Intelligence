## 6 · Scatter Plot

Now I will work with a (synthetic) dataset of insurance charges.

```python
# load dataset :
insurance = pd.read_csv(path+'seaborn_insurance.csv')

# show dataset :
insurance
```

---
Now I will see the relation of body mass index (BMI) with insurance charges by a **Scatter Plot**.

```python
# scatter Plot :
plt.figure(figsize=(14,7))
plt.title('BMI VS Charge')
sns.scatterplot(x=insurance.bmi, y=insurance.charges)
```

The scatterplot above suggests that 'body mass index' (BMI) and insurance charges are positively correlated, where customers with higher BMI typically also tend to pay more in insurance costs.

To double-check the strength of this relationship, you might like to add a regression line, or the line that best fits the data. We do this by changing the command to `sns.regplot`.

```python
# regrassion plot :
plt.figure(figsize=(14,7))
plt.title('BMI VS Charge')
sns.regplot(x=insurance.bmi, y=insurance.charges)
plt.xlabel('BMI')
plt.ylabel('Charge')
```

---
I can use scatter plots to display the relationships between (not two, but...) three variables! One way of doing this is by color-coding the points. Its called `Color-Coded scatter plot`.

For instance, to understand how smoking affects the relationship between BMI and insurance costs, we can color-code the points by 'smoker', and plot the other two columns ('bmi', 'charges') on the axes.

```python
# color coded scatter plot :
plt.figure(figsize=(14,7))
plt.title('BMI VS Charge')
sns.scatterplot(x=insurance.bmi, y=insurance.charges, hue=insurance.smoker)
plt.xlabel('BMI')
plt.ylabel('Charge')
```

This scatter plot shows that while nonsmokers to tend to pay slightly more with increasing BMI, smokers pay MUCH more.

---
To further emphasize this fact, we can use the `sns.lmplot()` command to add two regression lines, corresponding to smokers and nonsmokers. (The regression line for smokers has a much steeper slope, relative to the line for nonsmokers!)

```python
# lm plot :
sns.lmplot(data=insurance, x='bmi', y='charges', hue='smoker')
```

The `sns.lmplot` command above works slightly differently than the commands you have learned about so far:

- Instead of setting `x=insurance['bmi']` to select the 'bmi' column in insurance_data, we set `x="bmi"` to specify the name of the column only.
- Similarly, `y="charges"` and `hue="smoker"` also contain the names of columns.
- We specify the dataset with `data=insurance`.

----
Finally, there's one more plot that I will learn about.

Usually, we use scatter plots to highlight the relationship between two continuous variables (like "bmi" and "charges"). However, we can adapt the design of the scatter plot to feature a categorical variable (like "smoker") on one of the main axes.

We'll refer to this plot type as a categorical scatter plot, and we build it with the `sns.swarmplot()` command.

```python
sns.swarmplot(x=insurance['smoker'], y=insurance['charges'])
```

Among other things, this plot shows us that:

- on average, non-smokers are charged less than smokers, and
- the customers who pay the most are smokers; whereas the customers who pay the least are non-smokers.

## 7 · Histogram

Now I will work with a dataset of 150 different flowers, or 50 each from three different species of Iris (Iris setosa, Iris versicolor, and Iris virginica). Then I used it to make a **Histogram**. **Histogram** shows us **Distribution** of Data.


```python
# loading Iris Datasets :
iris = pd.read_csv(path+'seaborn_iris.csv', index_col='Id')

# Dataset :
iris
```

Each row in the dataset corresponds to a different flower. There are four measurements: the sepal length and width, along with the petal length and width. We also keep track of the corresponding species.

---
Now I will create a histogram to see how "Petal Length" varies in iris flowers. I can do this with the `sns.histplot` command.

```python
# histogram :
sns.histplot(iris['Petal Length (cm)'])
```

---
We can create three different histograms (one for each species) of petal length by using the `sns.histplot` command .

- `data=` provides the name of the variable that we used to read in the data
- `x= sets` the name of column with the data we want to plot
- `hue=` sets the column we'll use to split the data into different histograms


```python
# Histograms for each species
sns.histplot(data=iris, x='Petal Length (cm)', hue='Species')

# Add title
plt.title("Histogram of Petal Lengths, by Species")
```

---
## 8 · KDE Plot

The next type of plot is a **Kernel Density Estimate** (KDE) plot. The KDE plots, It can be imagined as a smoothed histogram. **KDE Plot** is a kind of Density Plot.

To make a **KDE** plot, we use the `sns.kdeplot` command. Setting `fill=True` colors the area below the curve (and `data=` chooses the column we would like to plot).

```python
sns.kdeplot(data=iris['Petal Length (cm)'], fill=True)
```

---
We can also create a KDE plot for each species by using `sns.kdeplot`. The functionality for `data`, `x`, and `hue` are identical to when we used `sns.histplot` above. Additionally, we set `fill=True` to color the area below each curve.

```python
# KDE plots for each species
sns.kdeplot(data=iris, x='Petal Length (cm)', hue='Species', fill=True)

# Add title
plt.title("Distribution of Petal Lengths, by Species")
```

One interesting pattern that can be seen in plots is that the plants seem to belong to one of two groups, where **Iris versicolor** and **Iris virginica** seem to have similar values for *petal length* , while **Iris setosa** belongs in a category all by itself.

In fact, according to this dataset, we might even be able to classify any **iris** plant as **Iris setosa** (as opposed to **Iris versicolor** or **Iris virginica**) just by looking at the petal length: if the petal length of an *iris*  flower is less than 2 cm, it's most likely to be Iris setosa!

---
## 9 · 2D KDE plots

We're not restricted to a single column when creating a KDE plot. We can create a **Two-Dimensional (2D)** KDE plot with the `sns.jointplot` command.

In the plot below, the color-coding shows us how likely we are to see different combinations of sepal width and petal length, where darker parts of the figure are more likely.

```python
# 2D KDE plot
sns.jointplot(x=iris['Petal Length (cm)'], y=iris['Sepal Width (cm)'], kind="kde")
```

Note that in addition to the 2D KDE plot in the center,
- the curve at the top of the figure is a KDE plot for the data on the x-axis (in this case, iris `['Petal Length (cm)']`),
- the curve on the right of the figure is a KDE plot for the data on the y-axis (in this case, iris `['Sepal Width (cm)']`).

# Conclusion

I have learned how to create many different chart types. Now, I will organize my knowledge.

Since it's not always easy to decide how to best tell the story behind any data, we have broken the chart types into **Three Broad Categories** to help with this :

- **Trends** - A trend is defined as a pattern of change.

  - `sns.lineplot` - **Line charts** are best to show trends over a period of time, and multiple lines can be used to show trends in more than one group.
 
---
- **Relationship** - There are many different chart types that we can use to understand relationships between variables in data.

  * `sns.barplot` - **Bar charts** are useful for comparing quantities corresponding to different groups.

  * `sns.heatmap` - **Heatmaps** can be used to find color-coded patterns in tables of numbers.

  * `sns.scatterplot` - **Scatter plots** show the relationship between two continuous variables; if **color-coded**, we can also show the relationship with a third categorical variable.

  * `sns.regplot` - Including a **regression line** in the scatter plot makes it easier to see any linear relationship between two variables.

  * `sns.lmplot` - This command is useful for drawing **multiple regression lines**, if the scatter plot contains multiple, color-coded groups.

  * `sns.swarmplot` - **Categorical scatter plots** show the relationship between a continuous variable and a categorical variable.


---
- **Distribution** - We visualize distributions to show the possible values that we can expect to see in a variable, along with how likely they are.


  - `sns.histplot` - **Histograms** show the distribution of a single numerical variable.

  - `sns.kdeplot` - **KDE plots** (or **2D KDE plots**) show an estimated, smooth distribution of a single numerical variable (or two numerical variables).

  - `sns.jointplot` - This command is useful for simultaneously displaying a **2D KDE plot** with the corresponding KDE plots for each individual variable.
