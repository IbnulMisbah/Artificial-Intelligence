## 2 · DataFrame

**Pandas** primarily works with two data structures: **Series** and **DataFrames**. A **DataFrame** is like a spreadsheet, with rows and columns. 

1. Series : a one-dimensional labeled array holding data of any type such as integers, strings, Python objects etc.
2. DataFrame : a two-dimensional data structure that holds data like a two-dimension array or a table with rows and columns.

Here’s how to create a simple DataFrame:

```python
# creating a DataFrame :
# firstly, making a `Dictionary` with `List` values:
data = {
    'Name' : ['Abdullah'],    # value must be a `List` / `Array`.
    'Age'  : [18],
}

# secondly, convarting Dictionary to DataFrame :
df = pd.DataFrame(data)

print(df, '\n')
print(type(df))
```

Here's `0` is called `Index`. By default, Pandas shows index as integer. like, 0, 1, 2, and so on. But I can change it with `index=` arguments.

---
Another DataFrame with `Index` :

```python
# a dataframe :
data = { 'Name' : [ 'Saem', 'Siam', 'Maruf', 'Atik', 'Mamun' ],
         'Grade': [ 'B', 'A', 'A+', 'B', 'A' ], }
indexes = ['B', 'A', 'A', 'B', 'A']

result = pd.DataFrame(data, index=indexes)

print(result, '\n')
```

Now the `Index` column is showing my given indexes.

---
### DataFrame to Series

I can make a Series from DataFrame. Lets do it...

```python
# Dataframe to series :
series_data = pd.Series(result['Grade'])

# to make a list of column :
columns = result.columns.to_list()

print(columns)
print(series_data)
print(type(series_data))
```

---
### NumPy to DataFrame

I also can make DataFrame from NumPy array by just adding column names. I can see DataFrame's shape by writing `pd.shape`.

Lets do it...

```python
# numpy to dataframe :
numpy_array = np.array([[0, 3], [2, 6], [4, 12], [5, 15]])
columns_name = ['temp', 'activity']

print(numpy_array, '\n', numpy_array.shape, '\n')

dataframe = pd.DataFrame(data=numpy_array, columns=columns_name)

dataframe
```

**Pandas** makes it easy to add new data in a DataFrame :

### Mapping Data

A `map` is a term, borrowed from mathematics, for a function that takes one set of values and "maps" them to another set of values. In data science we often have a need for creating new representations from existing data, or for transforming data from the format it is in now to the format that we want it to be in later. Maps are what handle this work, making them extremely important for getting your work done!

---
There are two mapping methods that I will use often :

#### pd.map()

`map()` is the first, and slightly simpler one. For example, suppose that we wanted to multiply the `has_soul` column by 100 to make percentages, we can use it :

```python
# map() fucntions to maps :
train.has_soul.map(lambda x: x * 100 )
```

---
But without `map()` I can do this with **Pandas** builtins operations.

```python
# maping without maps :
train.has_soul * 100
```

But `map()` is more flexible and builtins are more speedy.

---
#### pd.apply()

The `map()` function should expect a single value from the Series and return a transformed version of that value. `map()` returns a new Series where all the values have been transformed by function.

`apply()` is the equivalent method if I want to transform a whole DataFrame by calling a custom method on each row.

Lets see it :

```python
train.has_soul.apply(lambda x: x * 100)
```

```python
train.idxmax()
```
