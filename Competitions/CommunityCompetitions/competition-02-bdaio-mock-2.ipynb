### Evaluation

Submissions are scored using the Macro F1 Score.

----
### Why not just Accuracy?

Accuracy counts what percentage of your predictions are correct. Simple enough. But it can be misleading when one class appears more often than others. If 90% of the flowers were Shurjo, a model that always guesses Shurjo would have 90% accuracy, without learning anything useful.

F1 score fixes this by looking at two things at once:

- Precision: Of all the times your model said "this is Shurjo," how many times was it actually right?
- Recall: Of all the actual Shurjo flowers, how many did your model correctly find?

F1 combines these two into one number using the harmonic mean. The harmonic mean is stricter than a regular average. If either precision or recall is very low, the F1 score drops sharply. You cannot hide a weakness in one by doing well in the other.

Macro F1 means we calculate F1 separately for each of the three species and then average them. This ensures your model performs well across all three flowers, not just the most common one.

A perfect score is 1.0. A completely random model would score around 0.33.
