## 1 · Introduction

**Matplotlib** is a comprehensive library for creating static, animated, and interactive visualizations. It was created by neurobiologist **John Hunter** to work with EEG data. It grew to be used and developed by many people in many different fields. **John**'s goal was that **Matplotlib** make easy things easy and hard things possible.

Lets import **Matplotlib** to  learn it...

```python
# importing matplotlib and numpy :
import matplotlib.pyplot as plt
import numpy as np

# dark background :
#plt.style.use('dark_background')

if bool(np and plt):
    print('Matplotlib loaded successfully')
```

---
## 2 · An Example

Now I will write an easy example of **Matplotlib** to see a line graph. Lets see it...

```python
# easy example of Matplotlib :
# a graph of (number, square) :

# creating (x, y) = (numbers, squares) :
numbers = [i for i in range(6)]
squares = [i**2 for i in range(6)]

# mapping plot with numbers and squares :
plt.plot(numbers, squares)

# showing the graph
plt.show()
```

---
## 3 · Decoration

Now I will decorate my simple **Matplotlib** to make it more beautiful.

```python
# making graph more beatiful :
x = [n for n in range(6)]
y = [n**2 for n in range(6)]

# specifying size of figure :
plt.figure(figsize=(14, 5))    # (width, hight)

plt.plot(x, y, color='green', marker='o', linestyle='--', linewidth=1)  # marker='o', it will bold the dots.

plt.xlabel('Numbers')    # label for x_axis
plt.ylabel('Squares')    # label for y_axis

plt.title('My Cute Graph')  # Title for graph

plt.grid(True, alpha=0.5)          # it will show grid in graph

plt.show()
```

---
## 4 · Line Graph

Line Graph helps us to understand the trend and gives us a clear views of data. Now I will see a line graph of current weathers of my city.


```python
# making a line graph for days and temperature :
temp = [25, 27, 26, 30, 24]
days = [1, 2, 3, 4, 5]

plt.figure(figsize=(14,5))
plt.plot(days, temp, color='red', marker='o', linestyle='--', label='Temperature', linewidth=1)

plt.title('Weather of My City')
plt.xlabel('Days')
plt.ylabel('Celsius')

plt.grid(True)
plt.legend()
plt.show()
```

---
Now I will make a graph for number, square and cube to see thier relationship.

```python
# making a line graph of number, square and cube :
numbers = [n for n in range(6)]
squares = [n**2 for n in range(6)]
cubes   = [n**3 for n in range(6)]

plt.figure(figsize=(14, 5))
plt.plot(numbers, numbers, color='green', marker='o', linestyle='--', label='Numbers', linewidth=1)
plt.plot(numbers, squares, color='blue', marker='o', linestyle='--', label='Squares', linewidth=1)
plt.plot(numbers, cubes, color='purple', marker='o', linestyle='--', label='Cubes', linewidth=1)

plt.title('Line of Numbers, Squares and Cubes')

plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.grid(True)
plt.legend()
plt.show()
```

---
Now I will make a **Line Graph** of Trigonometric Functions like `sin()`, `cos()` and `tan()`.

```python
# line graph for sin(), cos() and tan()
x = np.linspace(-100, 100, 50)  # change interval to see megic.
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

plt.style.use('default')
plt.figure(figsize=(14, 5))

plt.plot(x, y1, label='sin(x)', linestyle='-', linewidth=2)
plt.plot(x, y2, label='cos(x)', linestyle='-', linewidth=2)
plt.plot(x, y3, label='tan(x)', linestyle='-', linewidth=2)

plt.title('Trigonometric Functions')

plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.legend()
plt.grid(True)
plt.show()
```

## 5 · Bar Chart

Now I will learn how to draw 'Bar Chart' with **Matplotlib** by drawing my "screen time" in this week.

```python
# drawing a bar chart :
days = ['Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed', 'Thu']
screen_hours = [7.45, 6.25, 6.25, 9.5, 4.8, 2.45, 3.2]

# colors in list :
colors = ['red', 'blue', 'green', 'orange', 'purple']

plt.figure(figsize=(14, 5))
plt.bar(days, screen_hours, color=colors, edgecolor='black')

plt.title('My Screen Time In This Week')

plt.xlabel('Days')
plt.ylabel('Hours')

#plt.xticks(rotation=45)
#plt.tight_layout()
plt.grid(True, alpha=0.2)
plt.show()
```

---
## 6 · Histogram

Now I will learn how to draw a histogram by **Matplotlib**. I will see age distribution of a class.

```python
# drawing histogram :
ages = [13, 14, 15, 12, 16, 14, 14, 15, 13, 17, 14, 13, 15, 16, 12, 14, 13, 15, 14, 13, 12]

plt.figure(figsize=(14, 5))
plt.hist(ages, bins=5, color='lightgreen' , edgecolor='black', alpha=0.8)   # bins = division / range

plt.title('Age Distribution')

plt.xlabel('Age Range')
plt.ylabel('Students')

plt.grid(True)
plt.show()
```

----
Now I will make a histogram of **NumPy** random numbers.

```python
# histogram :
#random_num = np.random.randint(1, 100, size=100)
random_num = np.random.randn(1000)

plt.figure(figsize=(14, 5))
plt.hist(random_num, bins=100, color='skyblue', edgecolor='black', alpha=0.7)

plt.title('Random Number Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.grid(True, alpha=0.2)
plt.show()
```

## 7 · Scatter Plot

Now I will learn how to draw scatter plot by **Matplotlib** . Scatter Plot used to understand the correlation between some things.

```python
# Scatter plot of Study hours VS Exam Score :
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
exam_scores = [45, 50, 62, 70, 78, 85, 92, 98]

plt.figure(figsize=(14, 5))
plt.scatter(study_hours, exam_scores, color='purple', marker='*')

plt.title('Study-Hours VS Exam-Scores')

plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')

#plt.grid(True)

plt.show()
```

---
Lets make a more beautiful **Scatter Plot**.

```python
# making a best scatter plot :
x = np.random.rand(100) * 100
y = np.random.rand(100) * 100

colors = np.random.rand(100)
sizes  = np.random.rand(100) * 1000

plt.figure(figsize=(15, 6))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='viridis')

# colorbar
plt.colorbar(label='Color Value')

plt.title('Scatter Plot with Colors and Size')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.show()
```

---
## 8 · Pie Chart

Now I will learn how to draw a Pie Chart. Pie Chart shows percentages of any activities.

```python
# Pie chart of my daily activities :
activities = [ 'Sleep', 'Study', 'Pray', 'Play', 'ECA', 'Others' ]
hours = [ 6, 6, 2, 1, 8, 1]

colors = ['gold', 'lightblue', 'lightcoral', 'lightskyblue', 'purple', 'lightgreen']
explode = (0,0,0,0,0.05,0)

plt.figure(figsize=(10,10))
plt.pie(hours, labels=activities, colors=colors, explode=explode , autopct='%1.1f%%', startangle=90)

plt.title('My Daily Spend Time')

plt.show()
```
